```python
import pandas as pd
```

```python
funnel = pd.read_csv('funnel.csv.csv')
funnel
```

```python
order = ['session_start', 'view_item', 'add_to_cart', 'begin_checkout', 'purchase']
funnel = funnel.set_index('event_name').loc[order].reset_index()
```

```python
funnel['prev_users'] = funnel['users'].shift(1)
funnel['step_conversion_pct'] = funnel['users'] / funnel['prev_users'] * 100
funnel['drop_off_pct'] = 100 - funnel['step_conversion_pct']
```

```python
funnel['overall_pct'] = funnel['users'] / funnel['users'].iloc[0] * 100
```

```python
funnel = funnel.round(1).drop(columns=['prev_users'])
funnel.to_csv('funnel_clean.csv', index=False)
funnel
```
